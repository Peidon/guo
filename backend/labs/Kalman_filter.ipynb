Sequence modeling is aim to handle sequential data. In other words, assume there are a serials data $x_0$, … , $x_t$, they could be used to predict $x_{t+1}$.
According to earlier experience. If we are going to make a model for predicting $x_{t+1}$, we are supposed to assign parameters to each input value of $x_0$,…, $x_t$. Actually , that's impossible. Because we need consider about the time scales and the dynamic nature of random variables.
A classical dynamical system has an inner state at each time step. We use a variable $h$ representing the state.

At each time step, RNN is in an inner state.

$$
h_t = f(Ux_t + Wh_{t-1} + b)
$$

The equation above is recurrent, because the definition of $h$ at time t refers back to the same definition at time $t-1$ .That's why the RNN (Recurrent neural networks) have feedback loops, which is the big difference compared to the other neural networks.

I'd like to try to explain the mathematical principles behind RNN. As we all known, machine learning models aim to fit the training samples with their true labels. The essence of model training is estimation for parameters.
When we use a predictive log-likelihood training objective, such as the equation below, we train the RNN to estimate the conditional distribution of the next sequence element $y^{t}$ given the past inputs and the actual observed outputs. This may mean that we maximize the log-likelihood.

$$
\mathrm{Log}_{e} p\left( y^{t} | x_1,... x_t, y^{t-1},...,y^{1} \right)
$$

In this case, the outputs $y$ are conditionally independent given the sequence of $x$ values. When we do feed the actual $y$ values back to the network, the directed graphical model contains edges from all $y^i$ values in the past to the current $y^t$ value.

As a simple example, let us consider the case that RNN models only a sequence of scalar random variables $Y = {y_1, ... , y_t}$, with no additional inputs $x$ . The input at time step $t$ is simply the output at time step $t-1$ . The RNN then defines a directly graphical model over the $y$ variables.

Although we believe that all the past inputs should have an influence on the next element of the sequence, there are no direct connections between the $y_i$ from distant past and the current $y_t$ in the graphical model structure of RNNs.

It's similar to Hidden Markov Model, where the distribution of sequence elements is regarded as joint distribution of hidden states and observations. The graphical model structure of RNNs results from regarding the hidden units $h^t$ as random variables.Including the hidden units in the graphical model reveals that the RNNs provide an efficient parametrization of the joint distribution over observations.
