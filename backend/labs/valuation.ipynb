Discounted Cash Flow

$$
DCF = \sum_{t=1}^{n}\frac{FCF_t}{(1+r)^t} + \frac{TV}{(1+r)^n}
$$

$FCF_t$ = Free Cash Flow in period $t$  
$TV$ = Terminal Value  
${r}$ = The discount rate (WACC,Weighted Average Cost of Capital) 

$$
TV = \frac{FCF_n\times(1+g)}{r-g}
$$

$FCF_n$ = Free Cash Flow in the final year of the projection period.  
$g$ = Perpetual growth rate of the cash flows (typically matched to long-term GDP growth or inflation).

```python


# The Weighted Average Cost of Capital (WACC) for the gold mining industry in Australia ranges between 7.5 and 12.5%.
# Where a project falls within this spectrum depends entirely on its operational stage:
# - Operating Mines (Tier-1): 7.5% – 9.5% WACC (Highly cash-generative, lower risk).
# - Development-Stage Projects: 9.5% – 11.5% WACC (Higher risk due to permitting, capital expenditure, and construction).
# - Exploration Stage: 11.5% – 14.0%+
wacc = 0.1                              # Discount rate (10%)

# VAL Inputs
# Comet & South Emu mine restarts cost 21M
# estimate 2026 fcf = 67.7 - 21 = 46.7M
# 2027 fcf = 15M from Comet
# 2028 and later fcf = 30M from Comet and Reedy
# ((4000+) - 3448) * 0.96 ≈ 500
# 500 * 30k ≈ 15M
# 2026 - 2030 FCFs = 46.7, 15, 30, 30, 30
# tototal_shares = 540M
fcfs = [46.7, 15, 30, 30, 30]           # Projected FCF (M$) for next 5 years
terminal_value = 30                     # Value at year 2030
total_shares = 540                      # Total shares outstanding (M)

# NMG Inputs
# 55k ounces * 500 = 27.5M
# 2026 - 2030 FCFs = 202,27.5,27.5,27.5,27.5
# total_shares = 10864M
# fcfs = [202,27.5,27.5,27.5,27.5]           # Projected FCF (M$) for next 5 years
# terminal_value = 27.5                     # Value at year 2030
# total_shares = 10864                      # Total shares outstanding (M)



# Discount cash flows
discount_factors = [(1 + wacc) ** i for i in range(1, len(fcfs) + 1)]
pv_fcfs = [f / d for f, d in zip(fcfs, discount_factors)]

# Discount terminal value
pv_terminal = terminal_value / (1 + wacc) ** len(fcfs)

# Enterprise and Equity Value
enterprise_value = sum(pv_fcfs) + pv_terminal

print(f"Enterprise Value: {enterprise_value:.2f} M$")
```

```python
# Assuming 540M shares outstanding
price = enterprise_value/total_shares
print(f"Stock Price: ${price:.2f}")
```
